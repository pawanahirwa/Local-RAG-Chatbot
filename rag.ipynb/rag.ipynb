```python
from langchain_community.document_loaders import PyPDFLoader
from langchain_text_splitters import RecursiveCharacterTextSplitter
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.vectorstores import Chroma

# Load PDF
loader = PyPDFLoader("C:\\Users\\ahirw\\project1\\project1\\data\\MTP 2 draft report.pdf")
docs = loader.load()

# Split text
splitter = RecursiveCharacterTextSplitter(
    chunk_size=500,
    chunk_overlap=50
)

chunks = splitter.split_documents(docs)

print(f"Total chunks: {len(chunks)}")

# Embedding model
embedding = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# Create vector database
db = Chroma.from_documents(
    chunks,
    embedding,
    persist_directory="vectordb"
)

db.persist()

print("Embeddings created successfully!")
```

```python
from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.llms import Ollama

# Embedding model
embedding = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# Load vector DB
db = Chroma(
    persist_directory="vectordb",
    embedding_function=embedding
)

# Local model
llm = Ollama(model="gemma:2b")

# Question
query = "What is this document about?"

# Retrieve chunks
results = db.similarity_search(query, k=3)

context = "\n\n".join([r.page_content for r in results])

# Prompt
prompt = f"""
Answer only from the context below.

Context:
{context}

Question:
{query}
"""

# Generate answer
response = llm.invoke(prompt)

print(response)
```

```python
import gradio as gr

from langchain_community.vectorstores import Chroma
from langchain_community.embeddings import HuggingFaceEmbeddings
from langchain_community.llms import Ollama

# Load embedding model
embedding = HuggingFaceEmbeddings(
    model_name="sentence-transformers/all-MiniLM-L6-v2"
)

# Load vector database
db = Chroma(
    persist_directory="vectordb",
    embedding_function=embedding
)

# Load local LLM
llm = Ollama(model="gemma:2b")


def rag_chat(query):

    # Retrieve relevant chunks
    results = db.similarity_search(query, k=3)

    # Combine chunks
    context = "\n\n".join([r.page_content for r in results])

    # Prompt
    prompt = f"""
    Answer only from the provided document context.

    If answer is not found in context, say:
    'Answer not found in document.'

    Context:
    {context}

    Question:
    {query}
    """

    # Generate answer
    response = llm.invoke(prompt)

    return response


# Gradio Interface
demo = gr.Interface(
    fn=rag_chat,
    inputs=gr.Textbox(
        lines=2,
        placeholder="Ask question from your PDF..."
    ),
    outputs="text",
    title="Local RAG Chatbot",
    description="Ask questions from your PDF document using local AI."
)

demo.launch()
```
